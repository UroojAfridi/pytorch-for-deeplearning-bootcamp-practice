#One of the most common errors in deepl learning(shape errors)

PyTorch implements matrix multiplication functionality in the torch.matmul() method.

The main two rules for matrix multiplication to remember are:

1. The inner dimensions must match:

(3, 2) @ (3, 2) won't work

(2, 3) @ (3, 2) will work

(3, 2) @ (2, 3) will work

2. The resulting matrix has the shape of the outer dimensions:

(2, 3) @ (3, 2) -> (2, 2)

(3, 2) @ (2, 3) -> (3, 3)

Note: "@" in Python is the symbol for matrix multiplication.

```python
tensor @ tensor
```

```python
# the inner dimesions must match
# the resulting matrix has the shape of the outer dimesions

torch.matmul(torch.rand(2,3) , torch.rand(3,2))
```

```python
torch.matmul(torch.rand(3,3) , torch.rand(3,3))
```

```python
torch.matmul(torch.rand(3,10) , torch.rand(10,3))
```

```python
torch.matmul(torch.rand(10,2) , torch.rand(2,10)).shape
```

```python
# shapes for matrix multiplication

tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])


torch.matmul(tensor_A, tensor_B)

# torch.mm(tensor_A, tensor_B) is same
```

```python
tensor_A.shape, tensor_B.shape
```

```python
# to fix tensor shape issues , we can manipulate the shape of one of our tensors using a transpose.
# it switches the axes or dimensions of given tensor



```

```python
tensor_B.T , tensor_B.T.shape
```

```python
tensor_B , tensor_B.shape
```

```python
torch.mm(tensor_A, tensor_B.T).shape
```

```python
torch.mm(tensor_A, tensor_B.T)
```

```python
# The operation works when tensor_B is transposed
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}\n")
print(f"New shapes: tensor_A = {tensor_A.shape} (same as above), tensor_B.T = {tensor_B.T.shape}\n")
print(f"Multiplying: {tensor_A.shape} * {tensor_B.T.shape} <- inner dimensions match\n")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput shape: {output.shape}")
```

```python
# rearrange the dim of target tensor in specified order

x_original = torch.rand(size=(224, 224, 3)) # [h,w,color_channel]

#permute the original tensor to a=rearrange the axis (or dim) order

x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"previous shape: {x_original.shape}")

print(f"New shape: {x_permuted.shape}") #color_channe;,h,w
```
