### Random Tensors

Random tensors are important because the way may neural network learn is that they strt with tesnors full of random nukbers and then adjust numbers to better represent the data.

start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers

```python
# create a random tensorof size or shape (3,4)

random_tensor = torch.rand(3, 4)

random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor.shape
```

```python
random_tensor = torch.rand(1, 3, 4)

random_tensor
```

```python
random_tensor.ndim
```

```python
# create a random tensor with similar shape to an image tensor

random_image_size_tensor = torch.rand(size=(224, 224, 3)) #height, width, colour channel
random_image_size_tensor.shape, random_image_size_tensor.ndim
```
