### Creating a range of tensors and tensor-like

```python
# use torch.range()

one_to_ten = torch.arange(start=0,end=11, step=1)
one_to_ten
```

```python
ten_zeros = torch.zeros_like(input=one_to_ten)
ten_zeros
```

### Tensor Datatypes

```python
# float 32 tensor

float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None,  #what dtype is tensor(e.g float32 or float 16)
                               device=None,  #what device is your tensor one
                               requires_grad=False)   # whether or not to track gradients wirth this tensor operations

float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)

float_16_tensor
```

```python
float_16_tensor * float_32_tensor
```

```python
int_32_tensor = torch.tensor([3, 6, 9], dtype=torch.int32)

int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

```python
## getting information from tensors

# Create a tensor
some_tensor = torch.rand(3, 4)
some_tensor
```

```python
# Find out details about it
print(some_tensor)
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Datatype of tensor: {some_tensor.dtype}")
print(f"Device tensor is stored on: {some_tensor.device}") # will default to CPU
```

### Manipulating Tensors (tensor operations)

```python
# create a tensor and add 10 to it

tensor = torch.tensor([1,2,3])

tensor + 10
```

```python
#multiply tensor by 10

tensor = tensor * 10
tensor
```

```python
#substract 10

tensor - 10
```

```python
# pytorch inbuilt functions

torch.mul(tensor, 10)
```

```python
torch.add(tensor,2)
```
